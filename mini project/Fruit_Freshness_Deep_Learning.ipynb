```python
import os
import zipfile
import requests
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

from tensorflow.keras import layers, models
from tensorflow.keras.applications import MobileNetV2
from sklearn.metrics import classification_report, confusion_matrix
from google.colab import files
from tensorflow.keras.utils import load_img, img_to_array

print("TensorFlow Version:", tf.__version__)

print("\nDownloading dataset...")

url = "https://github.com/vebynovalisa/fruit-freshness-classification/raw/refs/heads/main/Dataset.zip"

zip_path = "/content/Dataset.zip"

response = requests.get(url)

with open(zip_path, "wb") as f:
    f.write(response.content)

print("Dataset downloaded successfully")

extract_path = "/content/fruit_dataset"

with zipfile.ZipFile(zip_path, "r") as zip_ref:
    zip_ref.extractall(extract_path)

print("Dataset extracted successfully")

print("\nChecking dataset folders...")

for root, dirs, file_list in os.walk(extract_path):
    if file_list:
        print(root, ":", len(file_list), "images")


def find_folder(base_path, folder_name):
    for root, dirs, file_list in os.walk(base_path):
        if os.path.basename(root).lower() == folder_name.lower():
            return root
    return None


train_path = find_folder(extract_path, "train")
valid_path = find_folder(extract_path, "valid")
test_path = find_folder(extract_path, "test")

print("\nTrain folder:", train_path)
print("Validation folder:", valid_path)
print("Test folder:", test_path)


IMG_SIZE = (224, 224)
BATCH_SIZE = 16


print("\nLoading training data...")

train_data = tf.keras.utils.image_dataset_from_directory(
    train_path,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    shuffle=True
)


print("\nLoading validation data...")

valid_data = tf.keras.utils.image_dataset_from_directory(
    valid_path,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    shuffle=False
)


print("\nLoading test data...")

test_data = tf.keras.utils.image_dataset_from_directory(
    test_path,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    shuffle=False
)


class_names = train_data.class_names

print("\nClasses:", class_names)


AUTOTUNE = tf.data.AUTOTUNE

train_data = train_data.prefetch(AUTOTUNE)
valid_data = valid_data.prefetch(AUTOTUNE)
test_data = test_data.prefetch(AUTOTUNE)


print("\nShowing sample images...")


plt.figure(figsize=(10, 8))

for images, labels in train_data.take(1):

    for i in range(min(9, len(images))):

        plt.subplot(3, 3, i + 1)

        plt.imshow(images[i].numpy().astype("uint8"))

        label = int(labels[i].numpy()[0])

        plt.title(class_names[label])

        plt.axis("off")

plt.tight_layout()
plt.show()


data_augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.1),
    layers.RandomZoom(0.1)
])


print("\n==============================")
print("        SIMPLE CNN MODEL")
print("==============================")


cnn = models.Sequential([

    layers.Input(shape=(224, 224, 3)),

    data_augmentation,

    layers.Rescaling(1./255),

    layers.Conv2D(
        32,
        3,
        activation="relu"
    ),

    layers.MaxPooling2D(),

    layers.Conv2D(
        64,
        3,
        activation="relu"
    ),

    layers.MaxPooling2D(),

    layers.Conv2D(
        128,
        3,
        activation="relu"
    ),

    layers.MaxPooling2D(),

    layers.Flatten(),

    layers.Dense(
        128,
        activation="relu"
    ),

    layers.Dropout(0.5),

    layers.Dense(
        1,
        activation="sigmoid"
    )
])


cnn.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)


cnn.summary()


print("\nTraining CNN...")


history_cnn = cnn.fit(
    train_data,
    validation_data=valid_data,
    epochs=15
)


cnn_loss, cnn_accuracy = cnn.evaluate(
    test_data,
    verbose=0
)


print("\nCNN Test Accuracy:",
      round(cnn_accuracy * 100, 2), "%")


print("\n==============================")
print("        MOBILENETV2 MODEL")
print("==============================")


base_model = MobileNetV2(
    weights="imagenet",
    include_top=False,
    input_shape=(224, 224, 3)
)


base_model.trainable = False


mobilenet = models.Sequential([

    layers.Input(shape=(224, 224, 3)),

    layers.Rescaling(
        1./127.5,
        offset=-1
    ),

    base_model,

    layers.GlobalAveragePooling2D(),

    layers.Dropout(0.3),

    layers.Dense(
        1,
        activation="sigmoid"
    )
])


mobilenet.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)


mobilenet.summary()


print("\nTraining MobileNetV2...")


history_mobilenet = mobilenet.fit(
    train_data,
    validation_data=valid_data,
    epochs=10
)


mobile_loss, mobile_accuracy = mobilenet.evaluate(
    test_data,
    verbose=0
)


print("\nMobileNetV2 Test Accuracy:",
      round(mobile_accuracy * 100, 2), "%")


print("\n==============================")
print("       MODEL COMPARISON")
print("==============================")


print(
    "CNN Accuracy        :",
    round(cnn_accuracy * 100, 2),
    "%"
)

print(
    "MobileNetV2 Accuracy:",
    round(mobile_accuracy * 100, 2),
    "%"
)


models_list = [
    "CNN",
    "MobileNetV2"
]


accuracy_list = [
    cnn_accuracy * 100,
    mobile_accuracy * 100
]


plt.figure(figsize=(7, 5))

bars = plt.bar(
    models_list,
    accuracy_list
)

plt.ylabel("Accuracy (%)")

plt.xlabel("Model")

plt.title("CNN vs MobileNetV2")

plt.ylim(0, 100)


for bar, value in zip(
    bars,
    accuracy_list
):

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value + 1,
        f"{value:.2f}%",
        ha="center"
    )


plt.show()


print("\n==============================")
print("      ACCURACY GRAPHS")
print("==============================")


plt.figure(figsize=(8, 5))

plt.plot(
    history_cnn.history["accuracy"],
    label="CNN Training"
)

plt.plot(
    history_cnn.history["val_accuracy"],
    label="CNN Validation"
)

plt.xlabel("Epoch")

plt.ylabel("Accuracy")

plt.title("CNN Training and Validation Accuracy")

plt.legend()

plt.show()


plt.figure(figsize=(8, 5))

plt.plot(
    history_mobilenet.history["accuracy"],
    label="MobileNetV2 Training"
)

plt.plot(
    history_mobilenet.history["val_accuracy"],
    label="MobileNetV2 Validation"
)

plt.xlabel("Epoch")

plt.ylabel("Accuracy")

plt.title(
    "MobileNetV2 Training and Validation Accuracy"
)

plt.legend()

plt.show()


print("\n==============================")
print("      MOBILENETV2 EVALUATION")
print("==============================")


y_true = []

y_pred = []


for images, labels in test_data:

    predictions = mobilenet.predict(
        images,
        verbose=0
    )

    predictions = (
        predictions > 0.5
    ).astype(int).flatten()

    y_true.extend(
        labels.numpy().astype(int)
    )

    y_pred.extend(
        predictions
    )


print("\nClassification Report:\n")


print(
    classification_report(
        y_true,
        y_pred,
        target_names=class_names,
        zero_division=0
    )
)


print("\nConfusion Matrix:\n")


cm = confusion_matrix(
    y_true,
    y_pred
)


print(cm)


plt.figure(figsize=(6, 5))

plt.imshow(cm)

plt.title(
    "MobileNetV2 Confusion Matrix"
)

plt.xlabel(
    "Predicted Label"
)

plt.ylabel(
    "Actual Label"
)


plt.xticks(
    range(len(class_names)),
    class_names
)

plt.yticks(
    range(len(class_names)),
    class_names
)


for i in range(len(cm)):

    for j in range(len(cm)):

        plt.text(
            j,
            i,
            cm[i, j],
            ha="center",
            va="center"
        )


plt.tight_layout()

plt.show()


print("\n==============================")
print("       TEST YOUR OWN IMAGE")
print("==============================")


uploaded = files.upload()


filename = list(
    uploaded.keys()
)[0]


original_img = load_img(
    filename
)


img = load_img(
    filename,
    target_size=(224, 224)
)


img_array = img_to_array(
    img
)


img_array = np.expand_dims(
    img_array,
    axis=0
)


prediction = mobilenet.predict(
    img_array,
    verbose=0
)[0][0]


if prediction > 0.5:

    result = "ROTTEN"

    confidence = prediction * 100

else:

    result = "FRESH"

    confidence = (1 - prediction) * 100


print("\nImage:", filename)

print(
    "Prediction:",
    result
)

print(
    "Confidence:",
    round(confidence, 2),
    "%"
)


plt.figure(figsize=(6, 6))

plt.imshow(original_img)

plt.axis("off")

plt.title(
    f"Prediction: {result}\n"
    f"Confidence: {confidence:.2f}%"
)

plt.show()
```
